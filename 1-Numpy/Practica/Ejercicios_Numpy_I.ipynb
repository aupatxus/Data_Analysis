## Ejercicio 8
Concatena los siguientes arrays verticalmente

```Python
a = np.arange(10).reshape(2,-1)
b = np.repeat(1, 10).reshape(2,-1)
```

````python

a = np.arange(10).reshape(2,-1)
b = np.repeat(1, 10).reshape(2,-1)

print("array a:\n", a, "a.shape", a.shape)
print("array b:\n", b, "b.shape", b.shape)


ejer8_array = np.vstack((a, b)) #Concatena los arrays verticalmente
print("\nLos arrays verticalmente concatenados con np.vstack((a, b))")
print(ejer8_array)
````

## Ejercicio 10
Encuentra todos los elementos en común entre ambos arrays. [Para ello usa la función `intersect1d`](https://numpy.org/doc/stable/reference/generated/numpy.intersect1d.html)

```Python
a = np.array([1,2,3,2,3,4,3,4,5,6])
b = np.array([7,2,10,2,7,4,9,4,9,8])
```

````python


# https://numpy.org/doc/stable/reference/generated/numpy.intersect1d.html

a = np.array([1,2,3,2,3,4,3,4,5,6])
b = np.array([7,2,10,2,7,4,9,4,9,8])

ejer10_array = np.intersect1d(a,b) #la función intersect1d Encuentra todos los elementos en común entre ambos arrays
print(ejer10_array)
````

## Ejercicio 20
Dado el siguiente array

```Python

```

2. ¿Cuántas dimensiones?
